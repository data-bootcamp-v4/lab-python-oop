## Challenge 1: Bank Account

Cree una clase BankAccount con los siguientes atributos y métodos:

Atributos:

- número_cuenta (un identificador único para la cuenta bancaria)
- saldo (el saldo actual de la cuenta. Por defecto, es 0)

Métodos:

- depósito(monto): agrega la cantidad especificada al saldo de la cuenta
- retirar(cantidad): resta la cantidad especificada del saldo de la cuenta
- get_balance() - devuelve el saldo actual de la cuenta
- get_account_number() - devuelve el número de cuenta de la cuenta

Instrucciones:

- Cree una clase BankAccount con los atributos y métodos anteriores.
- Pruebe la clase creando algunas instancias de BankAccount y realizando depósitos y retiros.
- Asegúrese de que el atributo número_cuenta sea único para cada instancia de CuentaBancaria.

*Sugerencia: cree un atributo de clase account_count. El atributo de clase account_count se utiliza para realizar un seguimiento del número total de cuentas bancarias que se han creado utilizando la clase BankAccount. Cada vez que se crea un nuevo objeto BankAccount, el atributo account_count se incrementa en uno. Esto puede resultar útil para diversos fines, como generar números de cuenta únicos o monitorear el crecimiento de la base de clientes de un banco.*

```python
"""BankAccount: un ejercicio pequeño de POO con clases, objetos y un atributo de clase."""
 
 
class BankAccount:
    # Atributo de clase: lo comparten todas las instancias y cuenta las cuentas creadas hasta ahora.
    account_count = 0
 
    def __init__(self, balance=0):
        # Incrementa el contador compartido y lo usa como número de cuenta único (1, 2, 3...).
        BankAccount.account_count += 1
        self.account_number = BankAccount.account_count
        self.balance = balance
 
    def deposit(self, amount):
        """Suma una cantidad positiva al saldo."""
        if amount <= 0:
            print("La cantidad a depositar debe ser positiva.")
            return
        self.balance += amount
 
    def withdraw(self, amount):
        """Resta una cantidad del saldo solo si hay fondos suficientes."""
        if amount <= 0:
            print("La cantidad a retirar debe ser positiva.")
            return
        if amount > self.balance:
            # En lugar de lanzar un error, avisamos y el saldo no cambia.
            print("Saldo insuficiente.")
            return
        self.balance -= amount
 
    def get_balance(self):
        """Devuelve el saldo actual de la cuenta."""
        return self.balance
 
    def get_account_number(self):
        """Devuelve el número de la cuenta."""
        return self.account_number
 
 
# Probando la clase BankAccount
# Creamos dos instancias con saldos iniciales de 1000 y 500
account1 = BankAccount(1000)
account2 = BankAccount(500)
 
print("Account 1 balance:", account1.get_balance())        # Debe imprimir 1000
print("Account 1 number:", account1.get_account_number())  # Debe imprimir 1
 
print("Account 2 balance:", account2.get_balance())        # Debe imprimir 500
print("Account 2 number:", account2.get_account_number())  # Debe imprimir 2
 
account1.deposit(500)   # Depositamos 500 en la primera cuenta
account1.withdraw(200)  # Retiramos 200 de la primera cuenta
print("Account 1 balance after transactions:", account1.get_balance())  # Debe imprimir 1300
 
account2.withdraw(600)  # Intentamos retirar 600 de la segunda cuenta
print("Account 2 balance after transactions:", account2.get_balance())  # Avisa de saldo insuficiente y sigue en 500
```

## Challenge 2: Savings Account

Create a SavingsAccount class that inherits from the BankAccount class. The SavingsAccount class should have the following additional attributes and methods:

Attributes:

- interest_rate (the annual interest rate for the savings account. By default - if no value is provided - sets it to 0.01)

Methods:

- add_interest() - adds the interest earned to the account balance
- get_interest_rate() - returns the interest rate for the account

Instructions:

- Create a SavingsAccount class that inherits from the BankAccount class and has the above attributes and methods.
- Test the class by creating a few instances of SavingsAccount and making deposits and withdrawals, as well as adding interest.

```python
class SavingsAccount(BankAccount):
    """Cuenta de ahorro: es una BankAccount que además genera intereses."""

    def __init__(self, balance=0, interest_rate=0.01):
        # super() llama al __init__ de la clase padre (BankAccount).
        # Así se reutiliza todo lo que ya hace: número de cuenta único y saldo inicial.
        super().__init__(balance)
        # Atributo nuevo, solo de las cuentas de ahorro.
        self.interest_rate = interest_rate

    def add_interest(self):
        """Calcula el interés sobre el saldo actual y lo suma a la cuenta."""
        interes = self.balance * self.interest_rate
        self.balance += interes

    def get_interest_rate(self):
        """Devuelve la tasa de interés de la cuenta."""
        return self.interest_rate


# Probando la clase SavingsAccount
# Cuenta de ahorro con 1000 de saldo y un 5 % de interés
ahorro1 = SavingsAccount(1000, 0.05)
# Cuenta de ahorro con 500 de saldo y la tasa por defecto (1 %)
ahorro2 = SavingsAccount(500)

print("Ahorro 1 número:", ahorro1.get_account_number())
print("Ahorro 1 tasa de interés:", ahorro1.get_interest_rate())    # Debe imprimir 0.05
print("Ahorro 2 número:", ahorro2.get_account_number())
print("Ahorro 2 tasa de interés:", ahorro2.get_interest_rate())    # Debe imprimir 0.01

# Métodos heredados de BankAccount: funcionan sin haberlos escrito de nuevo
ahorro1.deposit(500)     # 1000 + 500 = 1500
ahorro1.withdraw(200)    # 1500 - 200 = 1300
print("Ahorro 1 saldo tras movimientos:", ahorro1.get_balance())   # Debe imprimir 1300

ahorro1.add_interest()   # 1300 + 5 % (65) = 1365
print("Ahorro 1 saldo con intereses:", ahorro1.get_balance())      # Debe imprimir 1365.0

ahorro2.withdraw(600)    # No hay fondos: avisa y el saldo sigue en 500
ahorro2.add_interest()   # 500 + 1 % (5) = 505
print("Ahorro 2 saldo con intereses:", ahorro2.get_balance())      # Debe imprimir 505.0
```

### Challenge 3: Checking Account

Create a CheckingAccount class that inherits from the BankAccount class. The CheckingAccount class should have the following additional attributes and methods:

Attributes:

- transaction_fee (the fee charged per transaction. By default is 1$)
- transaction_count (the number of transactions made in the current month)

Methods:

- deduct_fees() - deducts transaction fees from the account balance based on the number of transactions made in the current month and the transaction fee per transaction. The method calculates the total fees by multiplying the transaction count with the transaction fee, and deducts the fees from the account balance if it's sufficient. If the balance is insufficient, it prints an error message. Otherwise, it prints how much it's been deducted. After deducting the fees, the method resets the transaction count to 0.
- reset_transactions() - resets the transaction count to 0
- get_transaction_count() - returns the current transaction count for the account

Instructions:

- Create a CheckingAccount class that inherits from the BankAccount class and has the above attributes and methods.
- Test the class by creating a few instances of CheckingAccount and making deposits, withdrawals, and transactions to deduct fees.

Note: To ensure that the transaction count is updated every time a deposit or withdrawal is made, we need to overwrite the deposit and withdraw methods inherited from the BankAccount class. 

```python
class CheckingAccount(BankAccount):
    """Cuenta corriente: es una BankAccount que cobra una comisión por cada movimiento."""

    def __init__(self, balance=0, transaction_fee=1):
        # Reutilizamos el constructor de BankAccount (número único y saldo inicial).
        super().__init__(balance)
        # Atributos nuevos, solo de las cuentas corrientes.
        self.transaction_fee = transaction_fee   # comisión por movimiento (1 $ por defecto)
        self.transaction_count = 0               # movimientos hechos este mes

    # --- Métodos sobrescritos ---
    # Los redefinimos para que, además de hacer lo de siempre, cuenten el movimiento.

    def deposit(self, amount):
        saldo_antes = self.balance
        super().deposit(amount)              # hace el depósito normal de BankAccount
        if self.balance != saldo_antes:      # solo cuenta si el depósito se realizó
            self.transaction_count += 1

    def withdraw(self, amount):
        saldo_antes = self.balance
        super().withdraw(amount)             # hace el retiro normal de BankAccount
        if self.balance != saldo_antes:      # solo cuenta si el retiro se realizó
            self.transaction_count += 1

    # --- Métodos nuevos ---

    def deduct_fees(self):
        """Cobra las comisiones del mes y reinicia el contador de movimientos."""
        total_comisiones = self.transaction_count * self.transaction_fee
        if total_comisiones > self.balance:
            print("Error: insufficient balance to deduct the transaction fees.")
        else:
            self.balance -= total_comisiones
            print(f"Transaction fees of {total_comisiones}$ have been deducted from your account balance.")
            self.reset_transactions()

    def reset_transactions(self):
        """Pone el contador de movimientos a 0."""
        self.transaction_count = 0

    def get_transaction_count(self):
        """Devuelve cuántos movimientos se han hecho este mes."""
        return self.transaction_count


# Probando la clase CheckingAccount
# Cuenta corriente con 1000 de saldo y comisión por defecto (1 $)
corriente1 = CheckingAccount(1000)

corriente1.deposit(500)    # 1000 + 500 = 1500  → movimiento 1
corriente1.withdraw(200)   # 1500 - 200 = 1300  → movimiento 2
corriente1.deposit(100)    # 1300 + 100 = 1400  → movimiento 3
print("Corriente 1 saldo:", corriente1.get_balance())                # Debe imprimir 1400
print("Corriente 1 movimientos:", corriente1.get_transaction_count()) # Debe imprimir 3

corriente1.deduct_fees()   # 3 movimientos × 1 $ = 3 $
print("Corriente 1 saldo tras comisiones:", corriente1.get_balance())  # Debe imprimir 1397
print("Corriente 1 movimientos:", corriente1.get_transaction_count())  # Debe imprimir 0

# Cuenta corriente con 5 de saldo y comisión de 2 $
corriente2 = CheckingAccount(5, 2)

corriente2.deposit(10)     # 5 + 10 = 15  → movimiento 1
corriente2.withdraw(12)    # 15 - 12 = 3  → movimiento 2
corriente2.withdraw(10)    # No hay fondos: no se hace y no cuenta
print("Corriente 2 saldo:", corriente2.get_balance())                # Debe imprimir 3
print("Corriente 2 movimientos:", corriente2.get_transaction_count()) # Debe imprimir 2

corriente2.deduct_fees()   # 2 × 2 $ = 4 $, pero solo hay 3 → error
print("Corriente 2 saldo:", corriente2.get_balance())               
```

```python
# Probando la clase CheckingAccount
# Creamos una cuenta corriente con 500 de saldo y una comisión de 2 $ por movimiento
cuenta = CheckingAccount(500, 2)

cuenta.deposit(100)     # 500 + 100 = 600  → movimiento 1
cuenta.withdraw(50)     # 600 - 50  = 550  → movimiento 2
cuenta.deduct_fees()    # 2 movimientos × 2 $ = 4 $ → 546
print("Current balance:", cuenta.get_balance())               # Debe imprimir 546
print("Transaction count:", cuenta.get_transaction_count())   # Debe imprimir 0

cuenta.deposit(200)     # 546 + 200 = 746  → movimiento 1
cuenta.withdraw(75)     # 746 - 75  = 671  → movimiento 2
cuenta.deduct_fees()    # 2 movimientos × 2 $ = 4 $ → 667
print("Current balance:", cuenta.get_balance())               # Debe imprimir 667
print("Transaction count:", cuenta.get_transaction_count())   # Debe imprimir 0
```
